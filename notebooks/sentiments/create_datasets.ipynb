```python
%load_ext autoreload
%autoreload 2

import sentiments_utils as utils
from datasets import load_dataset, Dataset, DatasetDict, concatenate_datasets
from typing import List, Dict
```

```python
SOURCE_COLUMN = "spa"
TARGET_COLUMN = "shp"
```

```python
# Dataset de pruebas
test_dataset = utils.load_parallel_dataset_text(path_es="americasnlp_datasets/dev.es", path_shp="americasnlp_datasets/dev.shp")
test_dataset = utils.dedupe_bilingual(test_dataset, langs=(SOURCE_COLUMN, TARGET_COLUMN))

# Dataset de entrenamiento y validación
train_val_dataset = utils.load_parallel_dataset_text(path_es="americasnlp_datasets/train.es", path_shp="americasnlp_datasets/train.shp")
train_val_dataset = utils.dedupe_bilingual(train_val_dataset, langs=(SOURCE_COLUMN, TARGET_COLUMN))
train_val_dataset = utils.remove_overlaps_bilingual(train_val_dataset, test_dataset, langs=(SOURCE_COLUMN, TARGET_COLUMN))
train_val_dataset = train_val_dataset.train_test_split(test_size=0.2, seed=42)
train_dataset = train_val_dataset['train']
val_dataset = train_val_dataset['test']
```

```python
# Save datasets to disk
train_dataset.to_json("parallel_datasets/train.jsonl", lines=True)
val_dataset.to_json("parallel_datasets/val.jsonl", lines=True)
test_dataset.to_json("parallel_datasets/test.jsonl", lines=True)
```
